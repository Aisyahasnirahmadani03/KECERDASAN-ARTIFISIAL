### **Prompt 1 Geolistrik Resistivitas**
### **(*Identifikasi Potensi Akuifer Dangkal Menggunakan Metode Geolistrik Resistivitas*)**

### **ROLE**
Kamu berperan sebagai seorang ahli geofisika yang memiliki keahlian dalam metode geolistrik resistivitas, khususnya dalam proses akuisisi, pengolahan, pemodelan, dan interpretasi data resistivitas untuk mengetahui kondisi bawah permukaan. 

### **TASK**
Selesaikan studi kasus identifikasi potensi akuifer dangkal menggunakan metode geolistrik resistivitas. Analisis permasalahan berdasarkan prinsip dasar metode geolistrik, hukum Ohm, resistivitas semu, konfigurasi elektroda, proses pengukuran, pengolahan data, hingga interpretasi kondisi bawah permukaan. Gunakan data yang diberikan untuk mengetahui kemungkinan keberadaan lapisan yang berpotensi menjadi akuifer serta jelaskan dasar geofisika yang digunakan dalam menentukan interpretasi tersebut.

### **CONTEXT**
Studi kasus dilakukan pada suatu daerah dataran rendah yang diperkirakan tersusun atas material aluvium seperti pasir, kerikil, lempung, dan batuan dasar. Daerah tersebut diduga memiliki potensi air tanah dangkal sehingga diperlukan metode geofisika untuk mengetahui variasi lapisan bawah permukaannya. Metode geolistrik resistivitas digunakan karena setiap material bawah permukaan memiliki sifat kelistrikan yang berbeda. Interpretasi harus mempertimbangkan pengaruh litologi, porositas, kandungan air, tingkat kejenuhan, dan kandungan lempung terhadap nilai resistivitas.

### **INPUT**
Gunakan studi kasus berupa survei geolistrik konfigurasi Schlumberger dengan panjang lintasan 300 meter. Data pengukuran menunjukkan nilai resistivitas semu sebesar 85 Ωm pada AB/2 sebesar 5 meter, 72 Ωm pada AB/2 sebesar 10 meter, 48 Ωm pada AB/2 sebesar 20 meter, 18 Ωm pada AB/2 sebesar 40 meter, 32 Ωm pada AB/2 sebesar 80 meter, dan 120 Ωm pada AB/2 sebesar 150 meter. Berdasarkan informasi awal, lapisan pasir atau kerikil yang jenuh air diperkirakan mempunyai resistivitas rendah hingga sedang, sedangkan lempung dapat menunjukkan resistivitas rendah dan batuan dasar yang lebih kompak dapat memiliki resistivitas lebih tinggi.

### **CONSTRAINTS**
 Jangan membuat data pengukuran tambahan yang tidak diberikan. Jangan menentukan jenis batuan hanya berdasarkan satu nilai resistivitas karena nilai tersebut dapat dipengaruhi oleh beberapa faktor. Pertimbangkan kemungkinan adanya ambiguitas antara lapisan lempung dan lapisan yang mengandung air. Jika interpretasi tidak dapat dipastikan berdasarkan data yang tersedia, nyatakan bahwa interpretasi tersebut membutuhkan verifikasi menggunakan data tambahan.

### **OUTPUT**
Berikan hasil akhir dalam bentuk narasi yang menjelaskan permasalahan penelitian, prinsip metode geolistrik resistivitas, proses pengukuran menggunakan konfigurasi Schlumberger, perubahan nilai resistivitas terhadap jarak elektroda, interpretasi kemungkinan lapisan bawah permukaan, kemungkinan keberadaan akuifer dangkal, serta hubungan antara nilai resistivitas dan kondisi geologi. Jelaskan pula kemungkinan interpretasi alternatif yang dapat muncul akibat keberadaan lempung atau perubahan litologi. Akhiri dengan kesimpulan mengenai kondisi bawah permukaan dan potensi akuifer berdasarkan data yang tersedia. 

---


```python
import numpy as np
import matplotlib.pyplot as plt

# ===== INPUT (sesuai data pada prompt; tidak ada data tambahan) =====
AB2   = np.array([5, 10, 20, 40, 80, 150], dtype=float)   # AB/2 (m)
rho_a = np.array([85, 72, 48, 18, 32, 120], dtype=float)  # resistivitas semu (Ohm.m)
panjang_lintasan = 300.0                                  # m

# ===== 1. Pemeriksaan geometri survei =====
AB_maks = 2 * AB2.max()
print("=" * 62)
print("1. PEMERIKSAAN GEOMETRI SURVEI")
print("=" * 62)
print(f"AB maksimum = 2 x {AB2.max():.0f} = {AB_maks:.0f} m  (panjang lintasan = {panjang_lintasan:.0f} m)")
print("Faktor geometri K = pi*[(AB/2)^2-(MN/2)^2]/MN -> MN TIDAK diberikan,")
print("sehingga K dan beda potensial (dV) tidak dihitung (tidak mengarang data).")

# ===== 2. Kemiringan kurva pada skala log-log =====
slope = np.diff(np.log10(rho_a)) / np.diff(np.log10(AB2))
print("\n" + "=" * 62)
print("2. TABEL DATA DAN KEMIRINGAN KURVA (skala log-log)")
print("=" * 62)
print(f"{'AB/2 (m)':>9} {'rho_a (Ohm.m)':>14} {'Perubahan thd titik sebelumnya':>32}")
for i in range(len(AB2)):
    if i == 0:
        print(f"{AB2[i]:9.0f} {rho_a[i]:14.0f} {'-':>32}")
    else:
        pct = (rho_a[i] - rho_a[i-1]) / rho_a[i-1] * 100
        print(f"{AB2[i]:9.0f} {rho_a[i]:14.0f} {pct:+10.1f} %  (kemiringan = {slope[i-1]:+.2f})")

# ===== 3. Ciri kurva =====
i_min = np.argmin(rho_a)
print("\n" + "=" * 62)
print("3. CIRI KURVA SOUNDING")
print("=" * 62)
print(f"Nilai minimum : {rho_a[i_min]:.0f} Ohm.m pada AB/2 = {AB2[i_min]:.0f} m")
print(f"Nilai awal    : {rho_a[0]:.0f} Ohm.m  | Nilai akhir : {rho_a[-1]:.0f} Ohm.m")
print(f"Turun {rho_a[0]:.0f} -> {rho_a[i_min]:.0f} (faktor {rho_a[0]/rho_a[i_min]:.1f}x), "
      f"lalu naik {rho_a[i_min]:.0f} -> {rho_a[-1]:.0f} (faktor {rho_a[-1]/rho_a[i_min]:.1f}x)")
print("Bentuk umum   : menurun lalu menaik (cekung) -> mengarah ke tipe H")
print("                (rho1 > rho2 < rho3): ada lapisan resistivitas rendah di tengah.")

# ===== 4. Perkiraan kasar kedalaman pengaruh (aturan praktis) =====
print("\n" + "=" * 62)
print("4. PERKIRAAN KASAR KEDALAMAN PENGARUH (aturan praktis 1/3 - 1/2 x AB/2)")
print("=" * 62)
for a, r in zip(AB2, rho_a):
    print(f"AB/2 = {a:5.0f} m -> kedalaman pengaruh ~ {a/3:5.1f} - {a/2:5.1f} m  (rho_a = {r:.0f} Ohm.m)")
print("Catatan: hanya orientasi kasar; kedalaman batas lapisan sebenarnya butuh inversi.")

# ===== 5. Uji kewajaran data terhadap model 1D berlapis =====
print("\n" + "=" * 62)
print("5. UJI KEWAJARAN DATA (cabang naik kurva Schlumberger maks. 45 derajat)")
print("=" * 62)
batas_naik = 1.0
for i, s in enumerate(slope):
    if s > batas_naik:
        rho_maks = rho_a[i] * (AB2[i+1] / AB2[i]) ** batas_naik
        print(f"Segmen AB/2 {AB2[i]:.0f} -> {AB2[i+1]:.0f} m: kemiringan {s:+.2f} > 1  (TIDAK wajar untuk 1D)")
        print(f"   rho_a teoretis maksimum pada AB/2 = {AB2[i+1]:.0f} m  ~ {rho_maks:.0f} Ohm.m, teramati {rho_a[i+1]:.0f} Ohm.m")
    elif s > 0:
        print(f"Segmen AB/2 {AB2[i]:.0f} -> {AB2[i+1]:.0f} m: kemiringan {s:+.2f} <= 1  (wajar)")

# ===== 6. Grafik =====
fig, ax = plt.subplots(1, 2, figsize=(14, 5.2))
fig.suptitle("Geolistrik Schlumberger - Identifikasi Potensi Akuifer Dangkal", fontsize=13, fontweight="bold")

a = ax[0]
for i, s in enumerate(slope):
    warna = "red" if s > batas_naik else "tab:blue"
    a.loglog(AB2[i:i+2], rho_a[i:i+2], "-", color=warna, lw=2.5)
a.loglog(AB2, rho_a, "o", color="k", ms=7, label="Data lapangan")
a.loglog([AB2[-2], AB2[-1]], [rho_a[-2], rho_a[-2] * (AB2[-1] / AB2[-2])], "--", color="gray",
         label="Batas kemiringan 45° (maks. wajar)")
a.annotate("Minimum 18 Ωm\n(AB/2 = 40 m)", xy=(40, 18), xytext=(9, 14),
           arrowprops=dict(arrowstyle="->"), fontsize=10)
a.annotate("Naik terlalu curam\n(kemiringan 2,1)", xy=(150, 120), xytext=(22, 100),
           arrowprops=dict(arrowstyle="->", color="red"), fontsize=10, color="red")
a.set_xlabel("AB/2 (m)"); a.set_ylabel("Resistivitas semu (Ωm)")
a.set_title("(a) Kurva sounding (merah = segmen di luar batas 1D)")
a.set_xlim(4, 200); a.set_ylim(10, 200)
a.grid(True, which="both", ls="--", alpha=0.4); a.legend(loc="upper left", fontsize=9)

b = ax[1]
mid = np.sqrt(AB2[:-1] * AB2[1:])
warna_bar = ["red" if s > batas_naik else ("tab:blue" if s < 0 else "tab:green") for s in slope]
b.bar(range(len(slope)), slope, color=warna_bar, edgecolor="k")
b.axhline(1, color="gray", ls="--"); b.axhline(0, color="k", lw=0.8)
b.text(-0.45, 1.07, "batas kemiringan naik = +1", ha="left", fontsize=9, color="gray")
b.set_xticks(range(len(slope)))
b.set_xticklabels([f"{AB2[i]:.0f}→{AB2[i+1]:.0f} m" for i in range(len(slope))])
for i, s in enumerate(slope):
    b.text(i, s + (0.08 if s > 0 else -0.22), f"{s:+.2f}", ha="center", fontsize=10)
b.set_xlabel("Segmen AB/2"); b.set_ylabel("Kemiringan log(ρa) terhadap log(AB/2)")
b.set_title("(b) Kemiringan tiap segmen kurva")
b.set_ylim(-1.9, 2.5)
b.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

### **Interpretasi Prompt 1 — Geolistrik**

Resistivitas semu turun dari 85 Ωm hingga minimum 18 Ωm pada AB/2 = 40 m, lalu naik lagi sampai 120 Ωm, sehingga kurvanya berbentuk cekung (tipe H). Hal ini terjadi karena makin besar AB/2, arus menembus makin dalam: penurunan nilai menandakan adanya lapisan konduktif di kedalaman menengah (bisa pasir/kerikil jenuh air atau lempung), sedangkan kenaikan di akhir menandakan lapisan yang lebih resistif di bawahnya, kemungkinan batuan dasar. Namun kenaikan terakhir (80→150 m) terlalu curam untuk model 1D (kemiringan 2,1, batas wajar 1), sehingga titik ini perlu diukur ulang. Kesimpulannya, lapisan konduktif tersebut berpotensi menjadi akuifer dangkal, tetapi belum pasti karena lempung juga memiliki resistivitas rendah, sehingga perlu diverifikasi dengan data bor.

### **Prompt 2 MASW (Multichannel Analysis of Surface Waves)**
### **(*Identifikasi Kedalaman Batuan Dasar Menggunakan Metode MASW*)**

### **ROLE**
Kamu berperan sebagai seorang ahli geofisika yang memiliki keahlian dalam metode Multichannel Analysis of Surface Waves atau MASW, khususnya dalam analisis gelombang permukaan, kurva dispersi, dan interpretasi kecepatan gelombang geser.

### **TASK**
Selesaikan studi kasus untuk mengetahui kondisi lapisan tanah dan kedalaman batuan dasar menggunakan metode MASW. Analisis data berdasarkan prinsip gelombang Rayleigh, dispersi gelombang, kurva dispersi, kecepatan fase, dan kecepatan gelombang geser atau Vs. Gunakan hasil inversi Vs untuk menginterpretasikan perubahan sifat material terhadap kedalaman dan menentukan kemungkinan posisi batuan dasar.

### **CONTEXT**
Studi kasus dilakukan pada suatu daerah yang akan digunakan untuk pembangunan infrastruktur sehingga diperlukan informasi mengenai kondisi dan kekakuan lapisan bawah permukaan. Kondisi bawah permukaan diperkirakan terdiri atas tanah permukaan, material yang lebih kompak pada kedalaman tertentu, dan batuan dasar. Metode MASW digunakan untuk memperoleh model kecepatan gelombang geser karena perubahan Vs dapat memberikan informasi mengenai perubahan kekakuan material bawah permukaan.

### **INPUT**
Gunakan data survei MASW dengan 24 geophone yang memiliki jarak antar geophone sebesar 2 meter sehingga panjang array mencapai sekitar 46 meter. Sumber gelombang yang digunakan adalah palu dan plat besi dengan sampling interval 1 milidetik dan waktu perekaman 2 detik. Hasil inversi menunjukkan bahwa lapisan pertama memiliki kedalaman 0–4 meter dengan Vs sebesar 180 m/s, lapisan kedua berada pada kedalaman 4–12 meter dengan Vs sebesar 320 m/s, lapisan ketiga berada pada kedalaman 12–25 meter dengan Vs sebesar 650 m/s, sedangkan lapisan keempat berada pada kedalaman lebih dari 25 meter dengan Vs sebesar 900 m/s.

### **CONSTRAINTS**
Gunakan prinsip MASW secara benar dan jangan membuat nilai Vs atau kedalaman tambahan yang tidak tersedia. Jangan secara langsung menyatakan suatu nilai Vs sebagai jenis batuan tertentu tanpa mempertimbangkan informasi geologi. Bedakan antara kecepatan fase pada kurva dispersi dan Vs hasil inversi. Jelaskan bahwa peningkatan Vs secara umum menunjukkan material yang semakin kaku atau kompak, tetapi interpretasi litologi tetap membutuhkan informasi pendukung.

### **OUTPUT**
Berikan hasil akhir dalam bentuk narasi yang menjelaskan prinsip MASW, proses propagasi gelombang Rayleigh, pembentukan kurva dispersi, proses inversi hingga memperoleh model Vs, kemudian interpretasikan perubahan Vs pada setiap kedalaman. Jelaskan lapisan yang relatif lunak, lapisan yang lebih kompak, kemungkinan posisi batuan dasar, serta alasan geofisika yang mendukung interpretasi tersebut. Akhiri dengan kesimpulan mengenai kondisi bawah permukaan berdasarkan model MASW. 

---

```python
import numpy as np
import matplotlib.pyplot as plt

# ===== INPUT (sesuai data pada prompt) =====
n_geo, dx = 24, 2.0                 # jumlah geophone, jarak antar geophone (m)
dt, T = 0.001, 2.0                  # sampling interval (s), lama rekaman (s)
batas = np.array([0, 4, 12, 25])    # kedalaman atas tiap lapisan (m)
Vs    = np.array([180, 320, 650, 900], dtype=float)   # m/s (lapisan 4: > 25 m)

# ===== 1. Parameter akuisisi =====
L = (n_geo - 1) * dx
f_nyq = 1 / (2 * dt)
df = 1 / T
print("=" * 64)
print("1. PARAMETER AKUISISI")
print("=" * 64)
print(f"Panjang array              : ({n_geo}-1) x {dx:.0f} = {L:.0f} m")
print(f"Frekuensi Nyquist          : {f_nyq:.0f} Hz | resolusi frekuensi : {df:.1f} Hz")
print(f"Panjang gelombang maks.    : ~ panjang array = {L:.0f} m (aturan praktis)")
print(f"Panjang gelombang min.     : ~ 2 x dx = {2*dx:.0f} m (batas aliasing spasial)")
print(f"Kedalaman investigasi maks.: ~ lambda_maks/2 = {L/2:.0f} m (batas atas), "
      f"~ lambda_maks/3 = {L/3:.0f} m (konservatif)")

# ===== 2. Tabel lapisan hasil inversi =====
tebal = np.diff(batas).astype(float)
print("\n" + "=" * 64)
print("2. MODEL Vs HASIL INVERSI")
print("=" * 64)
print(f"{'Lapisan':<8}{'Kedalaman (m)':<16}{'Tebal (m)':<11}{'Vs (m/s)':<10}{'Vs/Vs atas':<11}{'(Vs/Vs1)^2':<10}")
for i in range(4):
    kdl = f"{batas[i]}-{batas[i+1]}" if i < 3 else f"> {batas[i]}"
    tb = f"{tebal[i]:.0f}" if i < 3 else "tak terhingga"
    rasio = "-" if i == 0 else f"{Vs[i]/Vs[i-1]:.2f}"
    kaku = (Vs[i] / Vs[0]) ** 2
    print(f"{i+1:<8}{kdl:<16}{tb:<11}{Vs[i]:<10.0f}{rasio:<11}{kaku:<10.1f}")
print("Kolom terakhir = kekakuan relatif (G ~ rho*Vs^2) DENGAN ASUMSI densitas sama (indikatif).")

# ===== 3. Vs30 (kecepatan rata-rata waktu tempuh) =====
h = np.append(tebal, 30 - batas[3])
vv = Vs
t_total = np.sum(h / vv)
Vs30 = 30 / t_total
print("\n" + "=" * 64)
print("3. Vs30 (ASUMSI: Vs lapisan 4 = 900 m/s berlaku sampai 30 m)")
print("=" * 64)
for i in range(4):
    print(f"  lapisan {i+1}: h = {h[i]:4.0f} m, Vs = {vv[i]:4.0f} m/s -> t = h/Vs = {h[i]/vv[i]*1000:6.2f} ms")
print(f"Vs30 = 30 / sum(h/Vs) = 30 / {t_total:.5f} = {Vs30:.0f} m/s")
print("Kelas situs (rujukan SNI 1726:2019): 350-750 m/s = SC (tanah sangat padat/batuan lunak)")
print("-> hanya perkiraan dari model yang diberikan; perlu verifikasi dengan data bor/geologi.")

# ===== 4. Kurva dispersi PERKIRAAN (bukan data ukur) =====
# Pendekatan: kecepatan fase Rayleigh ~ 0.92 x Vs rata-rata pada kedalaman z = lambda/3
def vs_rata(z):
    sisa, waktu = z, 0.0
    for i in range(4):
        tb = (batas[i+1] - batas[i]) if i < 3 else 1e9
        pakai = min(sisa, tb)
        waktu += pakai / Vs[i]
        sisa -= pakai
        if sisa <= 0:
            break
    return z / waktu

lam = np.linspace(4, 70, 200)
vph = np.array([0.92 * vs_rata(l / 3) for l in lam])
f = vph / lam
print("\n" + "=" * 64)
print("4. KURVA DISPERSI PERKIRAAN (pendekatan, BUKAN hasil pengukuran/inversi)")
print("=" * 64)
for l in (6, 12, 24, 46):
    k = np.argmin(abs(lam - l))
    print(f"lambda = {lam[k]:5.1f} m -> f ~ {f[k]:5.1f} Hz, kecepatan fase ~ {vph[k]:5.0f} m/s")
print("Frekuensi tinggi -> lambda pendek -> hanya 'melihat' lapisan dangkal (Vph kecil).")
print("Frekuensi rendah -> lambda panjang -> 'melihat' lapisan dalam (Vph besar).")

# ===== 5. Grafik =====
fig, ax = plt.subplots(1, 3, figsize=(16, 5.6))
fig.suptitle("MASW - Profil Vs dan Posisi Perkiraan Batuan Dasar", fontsize=13, fontweight="bold")

a = ax[0]
z_plot = np.append(batas, 40)
for i in range(4):
    a.fill_betweenx([z_plot[i], z_plot[i+1]], 0, Vs[i], alpha=0.5,
                    color=["#d9b38c", "#c49a6c", "#9c7a54", "#6b5a45"][i])
    a.text(Vs[i] + 20, (z_plot[i] + z_plot[i+1]) / 2, f"Lap.{i+1}: {Vs[i]:.0f} m/s", va="center", fontsize=9)
xs = np.repeat(Vs, 2)
ys = np.array([z_plot[0], z_plot[1], z_plot[1], z_plot[2], z_plot[2], z_plot[3], z_plot[3], z_plot[4]])
a.plot(xs, ys, color="k", lw=2)
a.axhline(L / 2, color="red", ls="--", label=f"λmaks/2 = {L/2:.0f} m")
a.axhline(L / 3, color="orange", ls="--", label=f"λmaks/3 = {L/3:.0f} m")
a.set_ylim(40, 0); a.set_xlim(0, 1150)
a.set_xlabel("Vs (m/s)"); a.set_ylabel("Kedalaman (m)")
a.set_title("(a) Profil Vs hasil inversi"); a.legend(loc="lower right", fontsize=9)
a.grid(alpha=0.3)

b = ax[1]
b.plot(lam, vph, color="tab:blue", lw=2)
b.axvline(L, color="red", ls="--", label=f"Panjang array = {L:.0f} m")
b.axvspan(L, lam.max(), color="red", alpha=0.08)
b.text(L + 1, vph.min() + 8, "di luar jangkauan\narray (kurang andal)", fontsize=9, color="red")
b.set_xlabel("Panjang gelombang λ (m)"); b.set_ylabel("Kecepatan fase perkiraan (m/s)")
b.set_title("(b) Dispersi PERKIRAAN (pendekatan, bukan data ukur)")
b.legend(loc="upper left", fontsize=9); b.grid(alpha=0.3)

c = ax[2]
pasang = ["1→2\n(4 m)", "2→3\n(12 m)", "3→4\n(25 m)"]
rasio_vs = Vs[1:] / Vs[:-1]
bars = c.bar(pasang, rasio_vs, color=["#c49a6c", "#9c7a54", "#6b5a45"], edgecolor="k")
for bb, r in zip(bars, rasio_vs):
    c.text(bb.get_x() + bb.get_width() / 2, r + 0.03, f"{r:.2f}x", ha="center", fontsize=11)
c.set_ylabel("Rasio Vs lapisan bawah / atas"); c.set_ylim(0, 2.4)
c.set_title("(c) Kontras Vs pada tiap batas lapisan"); c.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

### **Prompt 3 GPR (graound Penetrating Radar)**
### **(*Identifikasi Utilitas Pipa Bawah Permukaan*)**

### **ROLE**

### **CONTEXT**

Pada posisi sekitar 65 meter ditemukan pola refleksi berbentuk hiperbola dengan amplitudo relatif kuat dan memiliki puncak pada waktu tempuh sekitar 25 ns. Pola hiperbola tersebut tidak ditemukan secara konsisten pada bagian lintasan lainnya. ### **CONSTRAINTS**
Jangan langsung menyatakan bahwa reflektor berbentuk hiperbola pasti merupakan pipa karena bentuk tersebut juga dapat dihasilkan oleh objek bawah permukaan lainnya seperti kabel, batu, atau benda buatan lainnya. Jangan menentukan jenis material pipa secara pasti hanya berdasarkan radargram.
```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import fftconvolve

# ===== INPUT (sesuai data pada prompt) =====
f_antena = 400e6          # Hz (400 MHz)
panjang = 150.0           # m
x0 = 65.0                 # posisi puncak hiperbola (m)
t0 = 25.0                 # waktu tempuh puncak (ns)
c = 0.2998                # kecepatan cahaya di udara (m/ns)

# ===== 1. Kedalaman sebagai fungsi permitivitas (kecepatan TIDAK diberikan) =====
# d = v * t0 / 2 ; v = c / sqrt(er)  -> er hanya SKENARIO, bukan data
skenario = {"er = 4  (material kering)": 4, "er = 9  (tanah lembap)": 9,
            "er = 16 (tanah sangat lembap)": 16, "er = 25 (lempung jenuh)": 25}
print("=" * 70)
print("1. KEDALAMAN OBJEK PADA t0 = 25 ns UNTUK BEBERAPA SKENARIO er")
print("=" * 70)
print(f"{'Skenario':<32}{'v (m/ns)':>10}{'Kedalaman (m)':>15}{'Lambda (m)':>12}{'Resolusi ~L/4 (m)':>19}")
for nama, er in skenario.items():
    v = c / np.sqrt(er)
    d = v * t0 / 2
    lam = v / (f_antena / 1e9)
    print(f"{nama:<32}{v:>10.4f}{d:>15.2f}{lam:>12.3f}{lam/4:>19.3f}")
print("-> kedalaman objek ~0,75 - 1,87 m, bergantung kecepatan (belum bisa dipastikan).")

# ===== 2. Koefisien refleksi antarmuka (nilai er tipikal, perlu verifikasi) =====
def RC(e1, e2):
    return (np.sqrt(e1) - np.sqrt(e2)) / (np.sqrt(e1) + np.sqrt(e2))

tanah = 9
objek = {"Rongga udara (er=1)": 1, "Pipa PVC (er~3)": 3, "Lempung jenuh (er=25)": 25,
         "Air/pipa terisi air (er=81)": 81}
print("\n" + "=" * 70)
print("2. KOEFISIEN REFLEKSI DARI TANAH LEMBAP (er = 9) KE OBJEK")
print("=" * 70)
for nama, e in objek.items():
    r = RC(tanah, e)
    pol = "polaritas sama dgn gelombang datang" if r > 0 else "polaritas terbalik"
    print(f"{nama:<30} R = {r:+.3f}  -> {pol}")
print("Logam: R mendekati -1 (refleksi hampir total) -> amplitudo sangat kuat.")
print("Nilai er tipikal hanya untuk ilustrasi; perlu verifikasi di lapangan.")

# ===== 3. Model hiperbola difraksi: t(x) = sqrt(t0^2 + (2(x-x0)/v)^2) =====
def hiperbola(x, v):
    return np.sqrt(t0**2 + (2 * (x - x0) / v) ** 2)

x = np.linspace(55, 75, 601)

# ===== 4. Radargram SINTETIK ilustratif (er = 9 dipilih hanya untuk ilustrasi) =====
dt = 0.1                      # ns
t = np.arange(0, 60, dt)
v9 = c / 3.0
def ricker(fc, dt, panjang=6.0):
    tw = np.arange(-panjang/2, panjang/2, dt)
    a = (np.pi * fc * tw) ** 2
    return (1 - 2 * a) * np.exp(-a)
w = ricker(0.4, dt)
rad = np.zeros((len(t), len(x)))
tx = hiperbola(x, v9)
amp = 1.0 / np.sqrt(1 + ((x - x0) / 4.0) ** 2)    # pelemahan sederhana menjauhi puncak
for j in range(len(x)):
    k = int(round(tx[j] / dt))
    if k < len(t):
        rad[k, j] = amp[j]
rad = fftconvolve(rad, w[:, None], mode="same", axes=0)

# ===== 5. Grafik =====
fig, ax = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("GPR 400 MHz - Identifikasi Utilitas Pipa Bawah Permukaan", fontsize=13, fontweight="bold")

a = ax[0, 0]
a.imshow(rad, aspect="auto", cmap="gray", extent=[x[0], x[-1], t[-1], 0],
         vmin=-0.6, vmax=0.6)
a.set_ylim(45, 0)
a.axvline(x0, color="lime", ls="--", lw=1)
a.set_xlabel("Posisi (m)"); a.set_ylabel("Waktu tempuh dua arah (ns)")
a.set_title("(a) Radargram SINTETIK ilustratif (bukan data lapangan)")

b = ax[0, 1]
for (nama, er), warna in zip(skenario.items(), ["tab:green", "tab:blue", "tab:orange", "tab:red"]):
    b.plot(x, hiperbola(x, c / np.sqrt(er)), color=warna, lw=2, label=nama)
b.plot(x0, t0, "k*", ms=14, label="Puncak teramati (65 m, 25 ns)")
b.set_ylim(60, 0); b.set_xlim(55, 75)
b.set_xlabel("Posisi (m)"); b.set_ylabel("Waktu tempuh (ns)")
b.set_title("(b) Bentuk hiperbola untuk beberapa kecepatan")
b.legend(fontsize=8, loc="upper right"); b.grid(alpha=0.3)

cc = ax[1, 0]
er_range = np.linspace(3, 30, 100)
cc.plot(er_range, c / np.sqrt(er_range) * t0 / 2, color="tab:purple", lw=2)
for nama, er in skenario.items():
    cc.plot(er, c / np.sqrt(er) * t0 / 2, "ko")
cc.set_xlabel("Permitivitas relatif (er)"); cc.set_ylabel("Kedalaman objek (m)")
cc.set_title("(c) Kedalaman pada t0 = 25 ns vs permitivitas"); cc.grid(alpha=0.3)

d = ax[1, 1]
nama_o = ["Rongga\nudara", "Pipa\nPVC", "Lempung\njenuh", "Air"]
rv = [RC(tanah, e) for e in objek.values()]
d.bar(nama_o, rv, color=["tab:red" if r > 0 else "tab:blue" for r in rv], edgecolor="k")
d.axhline(0, color="k", lw=0.8)
for i, r in enumerate(rv):
    d.text(i, r + (0.03 if r > 0 else -0.07), f"{r:+.2f}", ha="center")
d.set_ylabel("Koefisien refleksi dari tanah (er=9)"); d.set_ylim(-0.7, 0.7)
d.set_title("(d) Polaritas dan kekuatan refleksi tiap tipe objek"); d.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import PchipInterpolator

# ===== INPUT (sesuai data pada prompt) =====
panjang = 1500.0                       # m
twt = np.array([600.0, 950.0, 700.0])  # ms : kiri, tengah, kanan
# Posisi kiri/tengah/kanan = ASUMSI (0, 750, 1500 m); prompt tidak memberi koordinat pasti
pos = np.array([0.0, 750.0, 1500.0])
nama = ["Kiri", "Tengah", "Kanan"]

# ===== 1. Relief waktu tempuh =====
print("=" * 66)
print("1. RELIEF WAKTU TEMPUH REFLEKTOR UTAMA (domain waktu, BUKAN kedalaman)")
print("=" * 66)
for n, p, t in zip(nama, pos, twt):
    print(f"{n:<7} x ~ {p:6.0f} m  TWT = {t:5.0f} ms")
d1 = twt[1] - twt[0]; d2 = twt[2] - twt[1]; d3 = twt[2] - twt[0]
print(f"\nKiri  -> Tengah : {d1:+.0f} ms (makin dalam)")
print(f"Tengah -> Kanan : {d2:+.0f} ms (kembali dangkal)")
print(f"Kiri vs Kanan   : {d3:+.0f} ms (kanan sedikit lebih dalam dari kiri)")
s1 = d1 / (pos[1] - pos[0]) * 100
s2 = d2 / (pos[2] - pos[1]) * 100
print(f"\nGradien sayap kiri  : {s1:+.1f} ms per 100 m")
print(f"Gradien sayap kanan : {s2:+.1f} ms per 100 m")
print(f"Rasio |kiri|/|kanan| = {abs(s1)/abs(s2):.2f}  -> sayap kiri lebih curam (asimetris)")
print(f"Amplitudo cekungan (TWT pusat - rata-rata tepi) = {twt[1] - twt[[0,2]].mean():.0f} ms")

# ===== 2. Penafsiran geometri (tanpa konversi kedalaman) =====
print("\n" + "=" * 66)
print("2. YANG BISA DAN TIDAK BISA DISIMPULKAN")
print("=" * 66)
print("Bisa   : reflektor melengkung ke bawah di tengah -> bentuk cekungan/depresi (sinklin)")
print("Bisa   : reflektor bawah mengikuti bentuk ini -> perubahan ketebalan lateral")
print("Tidak  : kedalaman (m)  -> butuh model kecepatan, TIDAK diberikan")
print("Tidak  : jenis batuan / jebakan hidrokarbon -> butuh data sumur/atribut seismik")

# ===== 3. Interpolasi bentuk reflektor (ILUSTRASI, antar titik tidak diketahui) =====
xx = np.linspace(0, panjang, 400)
tt = PchipInterpolator(pos, twt)(xx)

# ===== 4. Grafik =====
fig, ax = plt.subplots(1, 2, figsize=(14, 5.5), gridspec_kw={"width_ratios": [1.7, 1]})
fig.suptitle("Seismik Refleksi 2D - Konfigurasi Lapisan Sedimen dan Struktur Cekungan",
             fontsize=13, fontweight="bold")

a = ax[0]
a.fill_between(xx, tt, 1200, color="#e8d9b5", alpha=0.6, label="Sedimen di bawah reflektor utama")
a.plot(xx, tt, "-", color="tab:blue", lw=2.5, label="Reflektor utama (interpolasi, ilustratif)")
a.plot(pos, twt, "ro", ms=9, label="Titik data (dari prompt)")
for n, p, t in zip(nama, pos, twt):
    a.annotate(f"{n}\n{t:.0f} ms", xy=(p, t), xytext=(p, t - 85 if n == "Tengah" else t - 70),
               ha="center", fontsize=10, arrowprops=dict(arrowstyle="->"))
a.annotate("", xy=(750, 950), xytext=(750, 650), arrowprops=dict(arrowstyle="<->", color="red"))
a.text(770, 790, "+350 ms\nrelatif ke kiri", color="red", fontsize=9)
a.set_ylim(1200, 400); a.set_xlim(-50, 1550)
a.set_xlabel("Jarak sepanjang lintasan (m)  [posisi asumsi]"); a.set_ylabel("TWT (ms)")
a.set_title("(a) Geometri reflektor utama pada domain waktu")
a.legend(loc="lower center", fontsize=8); a.grid(alpha=0.3)

b = ax[1]
lbl = ["Kiri→Tengah", "Tengah→Kanan", "Kiri vs Kanan"]
nil = [d1, d2, d3]
bars = b.bar(lbl, nil, color=["tab:red", "tab:green", "tab:gray"], edgecolor="k")
b.axhline(0, color="k", lw=0.8)
for bb, v in zip(bars, nil):
    b.text(bb.get_x() + bb.get_width() / 2, v + (12 if v > 0 else -35), f"{v:+.0f} ms", ha="center", fontsize=11)
b.set_ylabel("Selisih TWT (ms)"); b.set_ylim(-320, 420)
b.set_title("(b) Perubahan TWT antar bagian lintasan")
b.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

### **Prompt 5 Induced Polarization (IP)**
### **(*Identifikasi Zona Mineralisasi Lempung*)**

### **ROLE**
Bertindaklah sebagai seorang ahli geofisika dan dosen yang berpengalaman dalam metode Induced Polarization (IP), khususnya dalam interpretasi resistivitas dan chargeability untuk mengidentifikasi perubahan sifat fisik bawah permukaan. Gunakan prinsip-prinsip geolistrik secara ilmiah dan jelaskan interpretasi berdasarkan hubungan antara respons listrik batuan dengan kondisi material bawah permukaan.

### **TASK**
Analisis data metode Induced Polarization (IP) untuk mengidentifikasi zona yang diduga memiliki kandungan material lempung tinggi di bawah permukaan. Interpretasikan pola resistivitas dan chargeability secara terpadu untuk menentukan zona yang memiliki karakteristik berbeda dari batuan sekitarnya serta jelaskan kemungkinan kondisi geologi yang menyebabkan respons tersebut.

### **CONTEXT**
Survei dilakukan pada daerah perbukitan yang tersusun oleh batuan vulkanik dengan tingkat pelapukan yang bervariasi. Beberapa bagian daerah penelitian menunjukkan kondisi tanah yang relatif lembap dan dijumpai material halus pada permukaan. Metode IP digunakan untuk mengetahui variasi sifat kelistrikan bawah permukaan dan membantu membedakan zona dengan karakteristik material yang berbeda.

### **INPUT**
Data diperoleh menggunakan konfigurasi Wenner-Schlumberger pada lintasan sepanjang 500 meter dengan spasi elektroda 10 meter. Hasil inversi menunjukkan bahwa bagian barat lintasan memiliki resistivitas sekitar 200–400 Ωm dan chargeability 3–6 mV/V, bagian tengah memiliki resistivitas 20–70 Ωm dan chargeability 15–25 mV/V, sedangkan bagian timur memiliki resistivitas 100–250 Ωm dan chargeability 5–9 mV/V. Pengamatan lapangan menunjukkan bahwa material permukaan di sekitar bagian tengah lintasan lebih halus dan relatif lembap dibandingkan bagian lainnya.

### **CONSTRAINT**
Jangan langsung menyimpulkan bahwa nilai resistivitas rendah dan chargeability tinggi hanya disebabkan oleh lempung karena respons tersebut juga dapat dipengaruhi oleh kandungan air, porositas, tingkat pelapukan, mineral konduktif, dan kondisi fluida pori. Interpretasi harus mempertimbangkan hubungan antara resistivitas, chargeability, kondisi permukaan, dan konteks geologi. Jangan menentukan jenis mineral lempung tertentu apabila tidak tersedia data pendukung seperti analisis mineralogi atau data geologi tambahan.

### **OUTPUT**
Tuliskan hasil analisis dalam bentuk narasi paragraf yang menjelaskan prinsip metode IP, hubungan resistivitas dan chargeability dengan sifat material bawah permukaan, pola perubahan nilai pada lintasan, serta interpretasi zona tengah sebagai zona yang memiliki karakteristik berbeda dari lingkungan sekitarnya. Jelaskan kemungkinan keterkaitan nilai resistivitas rendah dan chargeability tinggi dengan keberadaan material lempungan, kondisi lembap, atau tingkat pelapukan yang lebih tinggi, kemudian bahas kemungkinan penyebab lain dan keterbatasan interpretasinya. Akhiri dengan kesimpulan mengenai zona yang paling berpotensi memiliki kandungan material lempung tinggi berdasarkan integrasi data IP dan kondisi lapangan.


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# ===== INPUT (sesuai data pada prompt) =====
panjang, spasi = 500.0, 10.0
zona = ["Barat", "Tengah", "Timur"]
rho_rng = {"Barat": (200, 400), "Tengah": (20, 70), "Timur": (100, 250)}   # Ohm.m
M_rng   = {"Barat": (3, 6),     "Tengah": (15, 25), "Timur": (5, 9)}       # mV/V
# Batas zona = ASUMSI sepertiga lintasan (prompt hanya menyebut barat/tengah/timur)
batas_x = [0, panjang / 3, 2 * panjang / 3, panjang]
warna = {"Barat": "tab:blue", "Tengah": "tab:red", "Timur": "tab:green"}

# ===== 1. Parameter survei =====
n_elektroda = int(panjang / spasi) + 1
print("=" * 70)
print("1. PARAMETER SURVEI")
print("=" * 70)
print(f"Konfigurasi Wenner-Schlumberger, panjang {panjang:.0f} m, spasi {spasi:.0f} m")
print(f"Jumlah elektroda = {panjang:.0f}/{spasi:.0f} + 1 = {n_elektroda}")

# ===== 2. Ringkasan statistik tiap zona (titik tengah/rata-rata geometrik rentang) =====
print("\n" + "=" * 70)
print("2. RINGKASAN TIAP ZONA (dari rentang pada prompt)")
print("=" * 70)
print(f"{'Zona':<8}{'rho (Ohm.m)':<14}{'rho tengah*':<13}{'M (mV/V)':<11}{'M tengah':<10}{'Indeks M/rho x1000':<20}")
hasil = {}
for z in zona:
    r1, r2 = rho_rng[z]; m1, m2 = M_rng[z]
    rg = np.sqrt(r1 * r2)          # rata-rata geometrik (sesuai sifat log resistivitas)
    mm = (m1 + m2) / 2
    idx = mm / rg * 1000
    hasil[z] = (rg, mm, idx)
    print(f"{z:<8}{f'{r1}-{r2}':<14}{rg:<13.1f}{f'{m1}-{m2}':<11}{mm:<10.1f}{idx:<20.1f}")
print("* rata-rata geometrik rentang. Indeks M/rho x1000 = analog 'metal factor' (indikatif).")

# ===== 3. Kontras zona tengah terhadap zona lain =====
rg_t, m_t, i_t = hasil["Tengah"]
print("\n" + "=" * 70)
print("3. KONTRAS ZONA TENGAH TERHADAP ZONA LAIN")
print("=" * 70)
for z in ("Barat", "Timur"):
    rg, mm, idx = hasil[z]
    print(f"vs {z:<6}: resistivitas {rg/rg_t:4.1f}x lebih RENDAH | chargeability {m_t/mm:4.1f}x lebih TINGGI "
          f"| indeks M/rho {i_t/idx:5.1f}x lebih besar")
print("Rentang zona tengah tidak tumpang tindih dengan zona lain pada kedua parameter:")
for p, rng_ in (("resistivitas", rho_rng), ("chargeability", M_rng)):
    t = rng_["Tengah"]
    tumpang = any(not (rng_[z][1] < t[0] or rng_[z][0] > t[1]) for z in ("Barat", "Timur"))
    print(f"  {p:<14}: {'ADA tumpang tindih' if tumpang else 'tidak tumpang tindih'}")

# ===== 4. Grafik =====
fig = plt.figure(figsize=(16, 9))
fig.suptitle("Induced Polarization (Wenner-Schlumberger) - Identifikasi Zona Material Lempung",
             fontsize=13, fontweight="bold")
gs = fig.add_gridspec(2, 2, width_ratios=[1.5, 1])

a1 = fig.add_subplot(gs[0, 0])
for i, z in enumerate(zona):
    r1, r2 = rho_rng[z]
    a1.add_patch(Rectangle((batas_x[i], r1), batas_x[i+1] - batas_x[i], r2 - r1,
                           color=warna[z], alpha=0.55, label=f"{z}: {r1}-{r2} Ωm"))
    a1.text((batas_x[i] + batas_x[i+1]) / 2, np.sqrt(r1 * r2), z, ha="center", va="center", fontweight="bold")
a1.set_yscale("log"); a1.set_xlim(0, panjang); a1.set_ylim(10, 600)
a1.set_ylabel("Resistivitas (Ωm)"); a1.set_title("(a) Resistivitas sepanjang lintasan (batas zona asumsi)")
a1.legend(fontsize=8, loc="upper right"); a1.grid(alpha=0.3, which="both")

a2 = fig.add_subplot(gs[1, 0], sharex=a1)
for i, z in enumerate(zona):
    m1, m2 = M_rng[z]
    a2.add_patch(Rectangle((batas_x[i], m1), batas_x[i+1] - batas_x[i], m2 - m1,
                           color=warna[z], alpha=0.55))
    a2.text((batas_x[i] + batas_x[i+1]) / 2, (m1 + m2) / 2, z, ha="center", va="center", fontweight="bold")
a2.set_xlim(0, panjang); a2.set_ylim(0, 30)
a2.set_xlabel("Jarak sepanjang lintasan (m)"); a2.set_ylabel("Chargeability (mV/V)")
a2.set_title("(b) Chargeability sepanjang lintasan"); a2.grid(alpha=0.3)

a3 = fig.add_subplot(gs[:, 1])
for z in zona:
    r1, r2 = rho_rng[z]; m1, m2 = M_rng[z]
    a3.add_patch(Rectangle((r1, m1), r2 - r1, m2 - m1, color=warna[z], alpha=0.5))
    a3.text(np.sqrt(r1 * r2), (m1 + m2) / 2, z, ha="center", va="center", fontweight="bold")
a3.set_xscale("log"); a3.set_xlim(10, 600); a3.set_ylim(0, 30)
a3.set_xlabel("Resistivitas (Ωm)"); a3.set_ylabel("Chargeability (mV/V)")
a3.set_title("(c) Diagram silang resistivitas - chargeability")
a3.annotate("Resistivitas rendah\n+ chargeability tinggi\n(zona anomali)", xy=(35, 20), xytext=(90, 26),
            arrowprops=dict(arrowstyle="->"), fontsize=10)
a3.grid(alpha=0.3, which="both")
plt.tight_layout()
plt.show()
```

### **Interpretasi Prompt 5 — Induced Polarization**

Zona tengah lintasan memiliki resistivitas paling rendah (20–70 Ωm) dan chargeability paling tinggi (15–25 mV/V) dibandingkan zona barat dan timur, dengan rentang nilai yang tidak tumpang tindih. Hal ini terjadi karena lempung dan material lembap menghantarkan arus dengan baik sehingga resistivitas rendah, sementara lempung juga menahan muatan ion sehingga chargeability naik, dan kondisi permukaan yang halus dan lembap di bagian tengah mendukung hal tersebut. Namun resistivitas rendah juga bisa disebabkan air, dan chargeability setinggi itu bisa juga berasal dari mineral logam seperti pirit. Jadi bagian tengah paling berpotensi kaya material lempung, tetapi perlu dicek dengan sampel batuan atau analisis mineral.
